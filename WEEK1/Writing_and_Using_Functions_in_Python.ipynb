```python

def greet():
    print("Hello! Welcome to Python Functions.")

# Calling the function
greet()

```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")

```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Manashwini")

```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Manashwini", "Ashwitha", "Jahnavi")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Seetha", age=18, city="hyderabad")
```

```python

def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")

```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 15:", square(15))

```
